```python
# 1. Reload original 'created_at' to recover from 'Unknown' and incorrect datetime formatting
raw_df = pd.read_csv("/content/Pakistan Largest Ecommerce Dataset.csv", usecols=["created_at"])
df["created_at"] = raw_df["created_at"]

# 2. Convert 'created_at' to datetime using the correct format (M/D/Y) and handle missing values
df["created_at"] = pd.to_datetime(df["created_at"], errors="coerce")

# 3. Filter out rows where 'created_at' is null to ensure accurate time analysis
df_time = df[df["created_at"].notnull()]

# 4. Perform Monthly Sales Analysis
monthly_sales = (
    df_time.groupby(
        df_time["created_at"].dt.to_period("M")
    )["grand_total"]
    .sum()
    .reset_index()
)

# 5. Convert Period to String so it can be plotted easily
monthly_sales["created_at"] = monthly_sales["created_at"].astype(str)
monthly_sales
```

```python
plt.figure(figsize=(14, 6))

plt.plot(
    monthly_sales["created_at"],
    monthly_sales["grand_total"],
    marker="o"
)

plt.xticks(rotation=90)
plt.title("Monthly E-Commerce Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

```python
fig = px.line(
    monthly_sales,
    x="created_at",
    y="grand_total",
    markers=True,
    title="Monthly E-Commerce Revenue"
)

fig.show()
```

```python
Q1 = df["grand_total"].quantile(0.25)
Q3 = df["grand_total"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["grand_total"] < lower_bound) |
    (df["grand_total"] > upper_bound)
]

outliers.shape
```
